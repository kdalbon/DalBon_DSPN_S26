# Alpha Event-Related Desynchronization During Reward Processing in Depression

## Abstract

   **Background**: Alterations in the brain's reward system may contribute to deficits in motivation and pleasure observed in depression. Event-related desynchronization (ERD) in the EEG alpha band is thought to reflect a shift in resource allocation away from internally focused processes toward salient external events, such as reward outcomes. We hypothesized that individuals with depression would exhibit reduced alpha ERD during reward processing.  
   
   **Methods**: EEG was recorded while individuals with Major Depressive Disorder (MDD; n = 113) and Healthy Control participants (HC; n = 91) completed a casino-style slot machine task (**Figure 1**). Alpha band power (9-12 Hz), indexing neural oscillatory magnitude, was compared as a function of group (HC vs MDD) and reward outcome (win vs. loss) (**Figures 2 + 3**). Associations between reward-related alpha suppression and the reward positivity (RewP) event-related potential were examined across participants. Additionally, an exploratory **principal components analysis** was conducted to characterize time-frequency dynamics (**Figure 4**). 

   **Results**: A signifcant Group X Reward Outcome interaction (p=0.01) emerged for alpha power between 485 and 675ms following reward feedback. Healthy controls exhibited greater posterior alpha suppression for wins relative to losses (Cohen's d = 1.03), an effect that was attenuated in individuals with MDD (Cohen's d = 0.70). Across all participants, greater alpha suppression (ERD) to winning outcomes was associated with larger RewP amplitudes (p<0.001; partial eta^2=0.07), with no significant group differences in this relationship (**Figure 5**).

   **Conclusions**: These findings indicate altered alpha ERD during reward processing in depression, characterized by a blunted pattern of greater alpha suppression to wins than losses. Moreover, the observed covariation between ERD and RewP across groups underscores the role of alpha oscillatory dynamics in reward outcome processing.  


## Introduction

   Major Depressive Disorder (MDD) is associated with profound disruptions in emotional experience, including reduced motivation and a diminished ability to feel pleasure (ie., anhedonia). A large body of research suggests that these symptoms are linked to dysfunction in the brain's reward system: the set of neural processes that help individuals anticipate, respond to, and learn from rewarding events. People with depression often show reduced behavioral and neural responses to positive outcomes, pointing to impairments in how rewards are processed. Understanding the neural mechanisms underlying these differences is essential for improving models of depression and informing treatment approaches. 

   Electroencephalography (EEG) is particularly well suited for studying reward processing because it captures brain activity with high temporal precision. One commonly studied neural signal (ie., event-related potential) is the reward positivity (RewP), a brief change in brain activity that occurs shortly after receiving feedback and is typically larger following rewards than losses. Prior work has consistently found that the RewP is smaller in individuals with depression, suggesting reduced sensitivity to rewarding outcomes. However, most studies have focused on these time-locked signals, which provide only one perspective on brain activity. 

   Additional insight can be gained by examining brain activity in terms of neural oscillations, or rhythmic patterns of activity that unfold over time. Of particular interest is activity in the alpha frequency range (9-12 Hz). Decreases in alpha power following an event (referred to in this project as event-related desynchronization [ERD]) are generally thought to reflect increased engagement with external information, such as task-relevant stimuli. In the context of reward processing, stronger alpha ERD may indicate that an individual is allocating more attention and processing resources to feedback, especially when it is meaningful or salient, such as winning a reward.

   Despite growing interest in these oscillatory dynamics, relatively little is known about how alpha ERD relates to reward processing in depression. In particular, it remains unclear whether individuals with MDD show reduced alpha ERD in response to rewards, and how these oscillatory responses relate to more established measures like RewP (Burkhouse et al. 2018; doi: 10.1016/j.jad.2017.03.061). Examining both types of neural signals within the same study may provide a more complete picture of reward processing, as they likely reflect related but distinct aspects of brain function. 

   The present project aimed to address these questions by examining alpha-band activity during reward processing in individuals with MDD and Healthy Control (HC) participants using a well-establised slot machine task (Fryer et al. 2021; doi: 10.1016/j.neuroimage.2021.117874). We tested the hypothesis that individuals with MDD would show reduced alpha ERD in response to reward outcomes, particularly wins compared to losses. We also examined whether individual differences in alpha ERD were associated with RewP amplitude across participants, and conducted exploratory analyses to further characterize patterns of brain activity over time and frequency. By integrating multiple measures of neural activity, this project seeks to clarify how reward processing is altered in depression.

## Data Cleansing

#### Dimensionality Reduction via PCA: 

##### Does PCA reveal distinct underlying neural components of reward-related alpha band dynamics that explain the blunted reward responsiveness observed in depression?

Because the full dataset is too large to run a PCA on, the following script subsets the data first:

 - **frequencies** up to 40 Hz (drops the high-frequency end) 
 - **time** from -200 to 1000 ms (the peri-stimulus window of interest)
 - **conditions** only AAA and AAB (drops AA* and AB*) 

It then reshapes the 5D array into a 2D matrix so PCA can work on it. Here, I'm essentially treating each (frequency × time) combination as a "feature" and each (channel × condition × subject) combination as an "observation." 

The PCA outputs are then reshaped back into interpretable dimensions:

 - **coeff:** freq x time x component -- the time-frequency pattern of each component
 - **score:** channel x condition x subject x component -- how strongly each component is expressed across the scalp, conditions, and subjects 

```matlab
%% Data cleansing code

% 	[ nFreq, nTime, nChan, nCond, nSubj ] = size( WT );
% 	[ coeff, score ] = pca( reshape( WT, [ nFreq*nTime, nChan*nCond*nSubj ] )', 'Centered', true );

	% focus on subset of data
	iFreq = fwt <= 40;
	iTime = t >= -200 & t <= 1000;
	iCond = ismember( condType(:,2), {'AAA', 'AAB'} );
	[ nFreq, nTime, nChan, nCond, nSubj ] = size( WT(iFreq,iTime,:,iCond,:) );
```

#### Workflow Specification

No data anamolies present.

## Statistical Analysis

#### Run Principal Components Analysis (PCA)

```matlab
	% Run PCA
	[ coeff, score ] = pca( reshape( WT(iFreq,iTime,:,iCond,:), [ nFreq*nTime, nChan*nCond*nSubj ] )', 'Centered', true );
```

#### Reshape PCA outputs back to original dimensions

```matlab
	% Reshape PCA outputs back to original dimensions
	nComp = size( score, 2 );
	score = reshape( score, [ nChan, nCond, nSubj, nComp ] );
	coeff = reshape( coeff, [ nFreq, nTime, nComp ] );
```

## PCA Interpretation

The principal components analysis identified a dominant latent pattern of EEG activity characterized by increased loading in the alpha frequency range from about 200-700 ms after feedback presentation, consistent with post-outcome alpha desynchronization linked to attention and reward processing. Scalp topography maps showed that this component was strongest over posterior electrodes and was more strongly expressed for reward outcomes than nonreward outcomes, particularly in HCs. MDD individuals showed a similar spatial pattern but reduced differentiation between reward conditions, suggesting preserved basic neural architecture with diminished responsiveness to reward. Overall, running this PCA provided a confirmation that altered alpha-band dynamics during reward feedback are a meaningful feature of depression-related reward dysfunction.

## Conclusions

Together, these findings indicate that depression is associated with altered neural processing of reward feedback, reflected in a blunted pattern of posterior alpha-band desynchronization to winning relative to losing outcomes. Healthy Controls showed stronger alpha suppression following rewards, whereas individuals with Major Depressive Disorder exhibited reduced sensitivity to reward outcome despite retaining a similar overall spatial pattern of activity. The positive association between greater reward-related alpha suppression and larger RewP amplitudes further suggests that oscillatory alpha dynamics and event-related reward signals reflect complementary aspects of reward responsiveness. In addition, **the principal components analysis independently identified a dominant posterior alpha component that differentiated reward outcomes and showed attenuated modulation in depression, providing data-driven support for our primary hypothesis**. Collectively, these results suggest that disrupted attentional and motivational engagement with rewarding events may represent an important neurophysiological mechanism underlying reward dysfunction in depression.
