## 1. Deriving the Maximum Likelihood Estimate for Simple Linear Regression (6 points)

Using the mean squared error (MSE) as your objective function (the thing you're trying to minimize when you fit your model) allows for a closed form solution to finding the maximum likelihood estimate (MLE) of your model parameters in linear regression. Let’s consider the simple, single predictor variable model, i.e. simple linear regression :  $Y= \beta_0 + \beta_1 X $.

a) Use algebra to show how you can expand out $MSE(\beta_0, \beta_1)$ to get from i to ii below.

> _i)_ $E[ (Y-(\beta_0 + \beta_1 X))^2]$

> _ii)_ $E[Y^2] -2 \beta_0E[Y]-2 \beta_1 Cov[X,Y]-2 \beta_1 E[X]E[Y]+ \beta_0^2 +2 \beta_0 \beta_1 E[X]+\beta_1^2 Var[X]+ \beta_1^2 (E[X])^2$






**Answer:**

> $MSE(\beta_0, \beta_1)$ = $E[ (Y-(\beta_0 + \beta_1 X))^2]$
> 
> $(Y-(\beta_0 + \beta_1 X))^2 = Y^2 - 2Y(\beta_0 + \beta_1 X) + (\beta_0 + \beta_1 X)^2$
> 
> $- 2Y(\beta_0 + \beta_1 X) = -2\beta_0 Y - 2\beta_1 XY$
>
> and
>
> $(\beta_0 + \beta_1 X)^2 = \beta_0^2 + 2\beta_0 \beta_1 X + \beta_1 ^2 X^2$
> 
> 
> then combine:
> $(Y-(\beta_0 + \beta_1 X))^2 = Y^2 - 2\beta_0 Y - 2\beta_1 XY + \beta_0^2 + 2\beta_0\beta_1 X + \beta_1 ^2 X^2$
> $E[Y^2] - 2\beta_0 E[Y] - 2\beta_1 E[XY] + \beta_0^2 + 2\beta_0\beta_1 E[X] + \beta_1^2E[X^2]$
> 
>
> we know:
>
> $E[XY] = Cov(X,Y) + E[X]E[Y]$
>
> and
> 
> $E[X^2] = Var(X) + (E[X])^2$
>
> 
> So,
>
> $E[Y^2] - 2\beta_0 E[Y] - 2\beta_1 (Cov(X,Y) + E[X]E[Y]) + \beta_0^2 + 2\beta_0\beta_1 E[X] + \beta_1^2 (Var(X) + (E[X])^2)$
>
> $ = E[Y^2] - 2\beta_0 E[Y] - 2\beta_1 Cov(X,Y) - 2\beta_1 E[X]E[Y] + \beta_0^2 + 2\beta_0\beta_1 E[X] + \beta_1^2 Var(X) + \beta_1^2 (E[X])^2$     

b) Prove that the MLE of $\beta_0$ is $E[Y]- \beta_1 E[X]$ by taking the derivative of _ii_ above, with respect to $\beta_0$, setting the derivative to zero, and solving for $\beta_0$.

**Answer:**

> what involves $\beta_0$: $-2\beta_0 E[Y] + \beta_0^2 + 2\beta_0\beta_1 E[X]$
>
> $\frac{\partial}{\partial \beta_0} (-2\beta_0 E[Y]) = -2E[Y]$,
> 
> $\frac{\partial}{\partial \beta_0} (\beta_0^2) = 2\beta_0$,
> 
> $\frac{\partial}{\partial \beta_0} (2\beta_0\beta_1 E[X]) = 2\beta_1 E[X]$
> 
>
> So,
> 
> $\frac{\partial MSE}{\partial \beta_0} = -2E[Y] + 2\beta_0 + 2\beta_1 E[X]$
>
> Set equal to zero & solve:
> 
> $-2E[Y] + 2\beta_0 + 2\beta_1 E[X] = 0$
> 
> $-E[Y] + \beta_0 + \beta_1 E[X] = 0$
> 
> $\beta_0 = E[Y] - \beta_1 E[X]$

c) Prove that the MLE for $\beta_1$ is $Cov[X,Y]/Var[X]$ by taking the derivative of equation _ii_ above, with respect to $\beta_1$, setting the derivative to zero, and solving for $\beta_1$. *Hint: after you've simplified / expanded a bit, plug in the solution for $\beta_0$ from part b.*

**Answer:**

> keep only $\beta_1$:
> $-2\beta_1 Cov[X,Y] - 2\beta_1 E[X]E[Y] + 2\beta_0\beta_1 E[X] + \beta_1^2 Var[X] + \beta_1^2 (E[X])^2$
> 
> $\frac{\partial}{\partial \beta_1} -2\beta_1 Cov[X,Y] = -2Cov[X,Y]$,

> $\frac{\partial}{\partial \beta_1} -2\beta_1 E[X]E[Y] = -2E[X]E[Y]$,

> $\frac{\partial}{\partial \beta_1} 2\beta_0\beta_1 E[X] = 2\beta_0E[X]$,

> $\frac{\partial}{\partial \beta_1} \beta_1^2 Var[X] = 2\beta_1 Var[X]$,

> $\frac{\partial}{\partial \beta_1} \beta_1^2 (E[X])^2 = 2\beta_1 (E[X])^2$
>
> So,
>
> $\frac{\partial MSE}{\partial \beta_1} = -2Cov[X,Y] - 2E[X]E[Y] + 2\beta_0 E[X] + 2\beta_1 Var[X] + 2\beta_1 (E[X])^2$
>
> $-2Cov[X,Y] - 2E[X]E[Y] + 2\beta_0 E[X] + 2\beta_1 Var[X] + 2\beta_1 (E[X])^2 = 0$
>
> $-Cov[X,Y] - E[X]E[Y] + \beta_0 E[X] + \beta_1 Var[X] + \beta_1 (E[X])^2 = 0$
>
> Plugging in $\beta_0 = E[Y] - \beta_1 E[X]$ :
>
> $-Cov[X,Y] - E[X]E[Y] + (E[X]E[Y] - \beta_1 (E[X])^2) + \beta_1 Var[X] + \beta_1 (E[X])^2 = 0$
>
> $-Cov[X,Y] + \beta_1 Var[X] = 0$
>
> $\beta_1 Var[X] = Cov[X,Y]$
>
> $\beta_1 = \frac{Cov[X,Y]}{Var[X]}$
